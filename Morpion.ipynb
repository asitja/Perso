## Morpion

```python
class MORPION :

    '''
    jeu de morpion tour par tour,
    MORPION(j1, j2),
    jouer_coup(ligne,colonne),
    après j1 -> j2 et j2 -> j1
    '''

    def __init__(self, j1, j2):     
        self.j1 = j1
        self.j2 = j2
        self.tab = [[0,0,0],[0,0,0],[0,0,0]]
        self.nbcoups = 0
        self.historique = []
        self.joueurC = 1
        assert j1 != j2
        
        
    def __repr__ (self) :
        R = {"Joueur 1" : self.j1 , "Joueur 2" : self.j2 , "Nombre de coups" : self.nbcoups}
        return str(R)
    
    
    def valeur(self,v) :        
        if v == 0 :
            return (' ')
        if v == 1 :
            return (self.j1)
        if v == 2 :
            return (self.j2)
    
    
    def ligneT(self,i):                        # i = ligne   et   j = colone
        tiret = '\n --------------- \n'          
        ligne = tiret + '| ' + str(i+1) + ' | '
        for j in range (3) :
            ligne += self.valeur(self.tab[i][j]) + ' | '     # faire la ligne complete
        return (ligne)
    
    
    
    def ligneTest (self) :
        
        for i in range (3) :
            tl = self.tab[i][0]
            if tl != 0 :                                                    # regarde pour chaque ligne
                if (self.tab[i][1] == tl) and (self.tab[i][2] == tl) :
                    return (tl)                        
                        
            tc = self.tab[0][i]
            if tc != 0 :                                                    # regarde pour chaque colonne
                if (self.tab[1][i] == tc) and (self.tab[2][i] == tc) :
                    return (tc)
                    
        td = self.tab[1][1]
        if td != 0 :                                                        # regarde pour chaque diagonale
            if (self.tab[0][0] == td) and (self.tab[2][2] == td) :
                return (td)
                
            if (self.tab[0][2] == td) and (self.tab[2][0] == td) :
                return (td)
            
        
                       
                    
                    
    def est_gagnant (self) :            
        
        Ltest = self.ligneTest()
        if Ltest == 1 :
            return (self.j1)
        if Ltest == 2 :
            return (self.j2)
        
            
    
    def __str__ (self) :
        
        T = '|   | A | B | C |' + self.ligneT(0) + self.ligneT(1) + self.ligneT(2) + '\n \n'  # 1ere ligne + ligne de ligneT
        G = self.est_gagnant()
        
        if (G == self.j1) or (G == self.j2) :
            T += ('\n' + str(G) + ' est gagnant \n')
        
        return (T)
    
    
    
    def jouer_coup (self,i,j) :             # i = ligne   et   j = colone
        '''jouer_coup(ligne,colonne)'''
        i -= 1
        if j == 'A' :                  # remplace les lettres en chiffre (pour colonne)
            j = 0
        elif j == 'B' :
            j = 1
        else :
            j = 2
        
        if self.tab[i][j] != 0 :
            return (-1)
        else :
            self.tab[i][j] = self.joueurC               # self.joueurC = 1  (j1)
            self.joueurC = 3 - self.joueurC             # self.joueurC = 3 - 1 = 2  (j2)
            self.nbcoups += 1
            
```

```python
a = 'A'
b = 'B'
c = 'C'

M = MORPION('O', 'X')
M.jouer_coup(1,a)
M.jouer_coup(2,b)
M.jouer_coup(3,c)
M.jouer_coup(3,a)
M.jouer_coup(1,c)
M.jouer_coup(1,b)
M.jouer_coup(2,c)

print(M)
M
```
